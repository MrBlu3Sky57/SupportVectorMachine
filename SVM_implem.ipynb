```python
# Imports
import numpy as np # For storing data and performing computations
import email # For parsing data
from email import policy
from pathlib import Path # For parsing folder all together
from matplotlib import pyplot as plt # For visualizations
import random # For any random generation
SEED = 57131605
random.seed(SEED) # Deterministic for testing 
```

```python
spam_path = Path("enron1/spam")
ham_path = Path("enron1/ham")

# -1 label for spam and 1 label for ham
parsed_emails = []

# Parse Spam path
for file in spam_path.glob("*.txt"):
    with open(file=file, mode="r", encoding="latin-1") as f:
        parsed_emails.append((-1, email.message_from_file(f, policy=policy.default)))

# Parse Ham path
for file in ham_path.glob("*.txt"):
    with open(file=file, mode="r", encoding="latin-1") as f:
        parsed_emails.append((1, email.message_from_file(f, policy=policy.default)))

random.seed(SEED)
random.shuffle(parsed_emails)
```
